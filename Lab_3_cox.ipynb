1. As the parcel rises, it cools. Since cool air can hold less moisture, water vapor is lost as it condenses out of the air. As such, the moist adiabatic rate approaches the dry adiabatic rate as the moisture content becomes lower. Since the dry adiabatic rate is higher, the moist adiabatic rate increases.  
2. At a fixed pressure, as temperature increases, the lapse rate decreases. Warmer moist adiabats have a lower lapse rate when compared to cooler moist adiabats at the same pressure. 
3. skewing alows for an easier visualization of soundings by essentially streaching and rotating the graph to fit a better size. As the skew becomes more and more negative, the area of interest becomes squished to the bottom right of the graph, and is distorted. At positive skew values, the sounding data can also leave the viewing window and become compressed to the left. A small magnitude negative value of skew provides space between the adiabats, allowing for easy interpretation.
